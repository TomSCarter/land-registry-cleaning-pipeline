```python

output_orig = pd.read_parquet('../data/processed/pp_epc_oxfordshire_2015_2025_v1.parquet')

print(output_orig.shape)
print(output_orig.dtypes)
```

```python

output_new = pd.read_parquet('../data/processed/pp_epc_oxfordshire_2015_2025_v2.parquet')

print(output_new.shape)
print(output_new.dtypes)
```

```python
pp = pp.set_index('deed_date')
monthly_total = pp.resample('ME')['unique_id'].agg('count')
print(monthly_total)
```

```python
output_orig = output_orig.set_index('deed_date')
output_orig_monthly_total = output_orig.resample('ME')['unique_id'].agg('count')
print(output_orig_monthly_total)
```

```python
output_new = output_new.set_index('deed_date')
output_new_monthly_total = output_new.resample('ME')['unique_id'].agg('count')
print(output_new_monthly_total)
```

```python
combined = pd.DataFrame(monthly_total)
combined = combined.rename(columns={'unique_id':'price_paid_set'})
combined['v1_2015-2025'] = output_orig_monthly_total
combined['v1_match_rate'] = round(100* combined['v1_2015-2025'] / combined['price_paid_set'], 2)
combined['v2_2008-2025'] = output_new_monthly_total
combined['v2_match_rate'] = round(100* combined['v2_2008-2025'] / combined['price_paid_set'], 2)
combined['delta_match_rate'] = combined['v2_match_rate'] - combined['v1_match_rate']
print(combined)
```

```python
fig, ax = plt.subplots()
sns.lineplot(combined['v1_match_rate'], ax=ax, label = 'EPCs 2015-2025')
sns.lineplot(combined['v2_match_rate'], ax=ax, label = 'EPCs 2008-2025')
ax.legend(loc='right')
ax.set_ylabel('Match rate (%)')
ax.set_xlabel('Date of sale')
ax.set_ylim(top=100, bottom=0)
ax.set_title('Adding 2008-2014 EPCs removes the early-year gap in match rate', pad=15)
plt.tight_layout()
plt.show()
```
