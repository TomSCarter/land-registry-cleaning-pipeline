```python
for n in [df_orig, df_new]:
    n['inspection_date'] = n['inspection_date'].astype('datetime64[ns]')
    n['lodgement_date'] = n['lodgement_date'].astype('datetime64[ns]')
    n['dayofweek'] = n['lodgement_date'].dt.dayofweek
    n['day'] = n['lodgement_date'].dt.day
    n['month'] = n['lodgement_date'].dt.month
    n['quarter'] = n['lodgement_date'].dt.quarter
    n['year'] = n['lodgement_date'].dt.year
```
